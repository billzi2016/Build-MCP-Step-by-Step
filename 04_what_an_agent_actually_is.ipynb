# 04 What an Agent Actually Is

## Agent 不是一个更会聊天的模型，而是一个任务运行时

如果说过去两年 AI 领域最被滥用的词有哪些，`Agent` 一定在里面。很多产品只要做了多轮对话、加了几把工具、能自动输出几步计划，就会把自己包装成 Agent。这个说法不是完全错，但通常把问题说浅了。

真正的问题不是“这个系统会不会连续说很多轮话”，而是：**它有没有一个围绕目标推进、围绕状态更新、围绕外部能力调度的运行机制。**

如果没有，那么它最多只是一个带工具能力的对话系统。如果有，它才开始接近一个真正意义上的 Agent Runtime。

这一章要做的事情，就是把 Agent 从营销话语里拿出来，重新还原成一个工程对象。重点不是讲一个抽象定义，而是讲清楚：一个系统到底具备哪些结构之后，才有资格被称作 Agent。

## 先给结论

这一章最核心的判断可以浓缩成一句话：

> Agent 不是模型的一种人格，也不是产品的一种包装，而是一个让模型在目标约束下持续做出动作选择、吸收反馈并推进任务的运行时结构。

这句话里有几个关键词不能丢：

- `目标约束`：没有目标，只有聊天
- `持续动作选择`：没有动作选择，只有回答生成
- `吸收反馈`：没有反馈回路，只有单向输出
- `推进任务`：没有状态推进，只有一轮一轮的语义表演

所以 Agent 的本质，不是“像人一样思考”，而是“像系统一样闭环”。

## 1. Agent 与聊天模型的根本差异，不在语气，而在任务结构

最常见的误解，是把 Agent 理解成一个更主动、更连续、更会做事的聊天机器人。但如果只从外观上区分，几乎注定会混淆。

一个普通聊天系统也可以表现出很多“像 Agent”的表象：

- 能记住前面说过的话
- 能列计划
- 能根据用户补充继续往下答
- 能调用一两个工具

但这些能力本身并不能构成 Agent。真正的差异在于系统内部是否存在一条明确的任务推进逻辑。

聊天系统的基本单位通常是“本轮输入 -> 本轮输出”。它当然可以参考历史，但历史更多是上下文材料，而不是被系统显式维护的任务状态。

Agent 系统则不同。它的基本单位更接近：

- 当前目标是什么
- 当前状态推进到哪一步
- 下一步该执行什么动作
- 动作结果如何改变状态
- 什么时候判定任务完成或失败

换句话说，聊天系统以回答为中心，Agent 系统以任务闭环为中心。回答当然仍然存在，但回答不再是唯一主角，它只是闭环中的一个表现层。

## 2. 什么系统才配叫 Agent

为了避免概念泛化，最有效的办法不是给出一个更花哨的定义，而是给出一组最低判定条件。

一个系统至少满足下面几条，才有资格被严肃地称作 Agent：

- 它有显式目标，而不是仅仅对输入做反应
- 它能在多步过程中维护和更新状态
- 它能在多个候选动作之间做出选择
- 它能调用外部能力，并把结果重新纳入决策
- 它有终止条件，而不是无限续写
- 它对失败具备最基本的处理策略

如果少了其中大部分，那么系统更像“带增强功能的回答器”，而不是 Agent。

这个标准很重要。因为一旦标准过低，几乎任何带有历史记录的聊天应用都能自称 Agent；但那样的标签没有任何分析价值。

## 3. Goal：没有目标，就没有 Agent

Agent 的第一个基础件不是模型，也不是工具，而是目标。很多系统之所以看起来总像在“闲聊式帮忙”，根源就在于目标没有被系统化表达。

目标至少有三个层次：

- 会话层目标：这轮任务要达成什么结果
- 过程层目标：当前阶段要解决什么子问题
- 输出层目标：最后要以什么格式、什么标准交付结果

如果目标只停留在用户原话里，模型当然可以尝试自己理解，但系统不会真正拥有一个可维护的任务中心。只有当 runtime 把目标当作状态的一部分显式维护，后续的规划、执行、回填和终止判断才有公共参照。

这也是为什么很多“Agent demo”看起来能跑，但一旦任务变复杂就开始飘。它们没有真正意义上的目标管理，只有一条越来越长的对话历史。

一个最低限度的任务状态，至少应该能把下面这些东西显式放进去：

- 主目标：分析岗位 JD，并输出候选人能力映射与面试关注点
- 子目标：读取岗位要求、提取能力维度、对照候选人经历、生成结构化结论
- 完成标记：当前任务是否已经收束

重点不在于状态对象长什么样，而在于这些信息不能只漂浮在作者脑子里或聊天记录里。

## 4. State：没有状态推进，就没有任务推进

目标定义的是“要去哪”，状态定义的是“现在在哪”。Agent 之所以不是单轮问答，核心就在于它不能每一步都像失忆一样重新开始，而要知道当前任务已经走到哪一段。

状态不一定复杂，但至少应该能承载几类信息：

- 当前主目标
- 已完成的步骤
- 已读取或已确认的外部事实
- 当前待决的问题
- 上一轮工具调用或资源读取的结果

一旦没有这层，系统就会退回到“每轮都重新理解一次问题”的模式。表面上像多轮，实质上只是长上下文下的连续单轮。

很多人会把 memory 和 state 混在一起。更准确的区分是：memory 解决的是“能不能保留信息”，state 解决的是“保留的信息如何服务于当前任务推进”。前者偏存储，后者偏控制。

## 5. Policy：Agent 的关键不是会不会想，而是怎么决定下一步

Agent 的第三个核心件是 policy。这个词在不同语境下含义会有差别，但在这里，可以先把它理解成“下一步动作选择机制”。

对一个任务型 Agent 来说，当前轮通常至少有几种可能动作：

- 直接给出部分回答
- 请求澄清
- 读取某个资源
- 调用某个工具
- 更新内部状态并继续规划
- 结束任务

所谓 Agent，不是每一步都非得规划得多复杂，而是系统必须存在一个稳定的方式，去判断当前轮更适合哪类动作。

在 LLM 系统里，这个 policy 很少是一个独立的规则引擎，更多是由几层东西共同构成：

- system prompt 给出总体原则
- 当前状态提供上下文约束
- 工具和资源描述提供动作候选集合
- 模型根据这些上下文生成下一步动作意图
- runtime 对这个意图进行解释和执行

换句话说，LLM Agent 的 policy 往往是“模型决策 + runtime 限制”的混合体，而不是纯模型或纯规则单独完成。

## 6. Planning、Acting、Observing：Agent 的基本循环

一旦把 Agent 看成运行时结构，最自然的视角就不再是“它会不会回答”，而是“它怎么循环”。

一个最小但完整的 Agent 循环，通常可以拆成三段：

- Planning：决定当前任务应该往哪个方向推进
- Acting：执行一个外部动作，或者产生一个可执行动作意图
- Observing：读取动作结果，并据此修正状态和后续决策

这三段并不意味着 Agent 必须像学术论文里那样显式写出 `Thought -> Action -> Observation`。真正重要的是系统有没有这种功能分工。

没有 planning，系统就容易到处乱试；没有 acting，系统就只剩口头分析；没有 observing，系统就无法把执行结果真正纳入下一轮决策。

一个最小 Agent loop 至少可以被写成下面这六步：

1. 根据目标和当前状态决定下一步方向。
2. 在回答、澄清、读 resource、调 tool 等动作之间做选择。
3. 通过 Runtime 执行这个动作。
4. 读取返回结果。
5. 更新状态。
6. 判断是否满足终止条件。

只要这六步里缺掉一半，系统就很难算是一个像样的 Agent。

## 7. Tools 和 Resources：Agent 的外部能力边界

一个没有外部能力的 Agent，并不是逻辑上不可能，但在大多数实际系统里价值有限。因为只靠模型参数和会话文本，任务空间很快就会遇到事实获取、计算执行、文档读取、环境交互等边界。

因此，Agent 几乎总是围绕某种外部能力系统存在。这里至少要区分两类东西：

- `tools`：可以执行动作，产生副作用或外部结果
- `resources`：可以读取上下文，为后续决策提供材料

这两类东西对 Agent 来说都重要，但职责不同。工具让 Agent 能做事，资源让 Agent 不必瞎猜。

而这也正是后面 MCP 重要的原因之一。因为如果工具和资源只是零散地挂在某个程序里，Agent 可以用，但系统难以扩展、难以治理、难以被其他宿主复用。

## 8. Memory 不是聊天记录，而是任务可持续性的基础设施

讨论 Agent 时，memory 经常被说得很玄，仿佛只要系统有了记忆，就自动高级起来。实际上，memory 的工程价值很具体：它决定一个系统能不能跨越当前上下文窗口和当前轮次限制，继续稳定地推进任务。

但 memory 本身不是目的。真正值得关注的是 memory 被拿来做什么。

常见的 memory 至少可以分成几类：

- 会话记忆：保留当前任务过程中的已知信息
- 工作记忆：保留最近几步决策和执行结果，服务于短程推理
- 长期记忆：保留跨任务可复用的事实、偏好或经验

很多系统把完整聊天记录一股脑塞回模型里，然后把这件事叫 memory。技术上当然也能工作，但这更像粗暴堆上下文，不是严肃的记忆治理。真正成熟的 Agent 会关心哪些信息该保留、如何压缩、什么时候回填、什么内容只服务当前任务而不应永久保留。

## 9. Termination Condition：不会停的系统，不叫 Agent，叫风险

Agent 的另一个关键点，是它必须知道什么时候应该停。这个问题在 demo 里常常被忽略，因为演示者往往默认“跑几步就结束”。但只要系统开始真的调工具、读资源、做多轮决策，终止条件就从一个可选项变成了核心安全机制。

常见终止条件包括：

- 目标已经满足
- 达到最大步数
- 连续多步没有获得新信息
- 工具调用失败达到阈值
- 当前状态表明需要用户补充信息

如果没有这些边界，Agent 很容易出现两种问题：

- 明明已经能给答案，还在无意义地继续调用能力
- 明明任务已经卡死，还在机械地尝试下一步

这就是为什么终止条件并不是一个“最后再加上的工程细节”，而是 Agent 是否可控的核心组成部分。

## 10. Error Recovery：失败处理不是补丁，而是 Agent 的必要结构

只要 Agent 涉及外部能力，它就一定会失败。工具会报错，参数会不合法，资源会不存在，用户目标会不完整，模型还会选错动作。一个不考虑失败的 Agent，不是理想化，而是没真正进入工程语境。

因此，Agent 至少要具备几类基础恢复机制：

- 输入校验失败时，能否请求补充信息
- 工具执行失败时，能否重试或改走别的路径
- 结果噪音过高时，能否先压缩再继续推理
- 多轮无进展时，能否主动终止并说明原因

这些机制不需要一开始就复杂，但必须存在。因为 Agent 不是“只在理想路径上工作”的系统，而是“面对偏差仍能维持任务结构”的系统。

## 11. 为什么 Agent 才是 MCP 的直接调用方

到这里，已经可以清楚回答一个后面会反复出现的问题：为什么说 Agent 是 MCP 的直接调用方，而不是模型或协议本身在“自动工作”。

原因其实很朴素。MCP 暴露的是能力面，不是任务面。它能告诉系统：这里有哪些 tool、哪些 resource、哪些 prompt 可以用；但它不会替系统决定当前任务要不要读 resource、先调哪个 tool、失败之后该不该重试、什么时候结束。

这些判断都属于任务推进逻辑，而任务推进逻辑正是 Agent Runtime 的职责。

更具体地说：

- 模型负责在当前上下文里生成下一步意图
- Agent 负责把这个意图解释成系统动作
- MCP 负责把外部能力以标准化方式暴露出来

这三者一旦混在一起，就会出现一种很常见的误读：仿佛 MCP 自己能完成任务，或者模型自己会去执行工具。事实上，真正把“能力列表”变成“任务执行链”的，只能是 Agent。

把职责拆开写，其实已经足够清楚：

- `model`：在当前上下文中生成下一步最合理的输出意图
- `agent runtime`：维护目标与状态，决定如何解释和执行这个意图
- `mcp layer`：标准化暴露 tools、resources、prompts 等外部能力

三者一旦混起来，后面关于能力边界、任务边界和失败归因的讨论都会变脏。

## 12. Single-Agent 与 Multi-Agent，不是“几个模型”这么简单

既然 Agent 是运行时结构，那么 single-agent 和 multi-agent 的区别，也就不该被粗暴理解成“开一个模型”还是“开几个模型”。真正的区别在于任务控制流如何组织。

single-agent 更像一个中心执行单元：

- 一个目标中心
- 一套状态管理
- 一个主决策循环
- 多种外部能力可调用

multi-agent 则意味着任务被拆成多个相对独立的决策单元，它们可能有不同角色、不同状态视角、不同能力边界，甚至彼此之间也需要协调协议。

从工程现实看，很多系统并不需要 multi-agent。把一个 single-agent runtime 做稳，通常比快速堆出多个角色更有价值。因为多 agent 带来的不是自动更强，而是更复杂的上下文同步、状态一致性、成本和调试问题。

## 13. 为什么 Agent 这个词会被滥用

Agent 之所以容易被过度使用，是因为它刚好站在两个叙事交汇点上：

- 从产品视角看，它听起来比 chatbot 更有行动力
- 从技术视角看，它确实比单轮调用更接近系统化任务执行

问题在于，很多场景只借用了前者的叙事红利，却没有真正承担后者的工程要求。于是一个能多轮对话的系统、一个能输出计划的系统、一个能调函数的系统，都可能被包装成 Agent。

这就是为什么本文一直强调要把 Agent 还原成结构问题。只有这样，后面在谈 MCP、Runtime、Evaluation 时，讨论才不会被营销语言带偏。

## 14. 从 Agent 走向 MCP 的自然过渡

一旦接受 Agent 是任务运行时这个判断，接下来就会遇到一个非常实际的问题：外部能力应该如何被组织。

如果 Agent 只调用两三个本地函数，问题看起来不大。但只要系统开始扩展，工具越来越多、文档来源越来越散、资源读取需求越来越复杂，能力接入就会迅速变成治理问题，而不再只是编码问题。

这也是为什么 MCP 会在这个位置出现。它不是为了替代 Agent，而是为了给 Agent 提供一个更标准、更清晰、更可扩展的外部能力层。Agent 解决的是任务推进，MCP 解决的是能力暴露。两者不是竞争关系，而是分工关系。

## 15. 本章结论

如果把这一章压缩成一组最关键的结论，可以保留下面这些：

- Agent 的本质不是一个更会聊天的模型，而是一个任务运行时。
- 目标、状态、策略、动作、反馈、终止条件和错误恢复，缺一大半都不能算严肃的 Agent。
- memory 不是聊天记录的同义词，state 也不是 memory 的同义词。
- planning、acting、observing 构成的是功能循环，不是表演式话术。
- MCP 暴露能力，Agent 调度能力；真正把能力变成任务执行链的，是 Agent Runtime。

下一章会继续往前推进，但视角会转一下：不是继续定义 Agent，而是回答一个更工程化的问题。既然 Agent 需要调用越来越多外部能力，为什么传统的 function calling 和零散 prompt 拼接最终会不够用，MCP 为什么会在这个阶段变得有必要。
